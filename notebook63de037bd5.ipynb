```python
%cd /kaggle/working/ondeberta_optuna

p = "tune.py"
s = open(p).read()

rep = {
 'trial.suggest_float("lr", 1e-5, 5e-5, log=True)':        'trial.suggest_float("lr", 2e-5, 5e-5, log=True)',
 'trial.suggest_float("head_lr", 1e-4, 2e-3, log=True)':   'trial.suggest_float("head_lr", 1e-4, 6e-4, log=True)',
 'trial.suggest_float("warmup_ratio", 0.0, 0.15)':         'trial.suggest_float("warmup_ratio", 0.06, 0.2)',
 'trial.suggest_categorical("onlstm_hidden", [128, 256, 384])': 'trial.suggest_categorical("onlstm_hidden", [128, 256])',
 'trial.suggest_int("onlstm_layers", 1, 2)':               '1',
 'trial.suggest_float("adv_epsilon", 0.1, 1.5)':           'trial.suggest_float("adv_epsilon", 0.1, 1.0)',
 'trial.suggest_categorical("batch_size", [16, 32])':      '32',
 'optuna.samplers.TPESampler(seed=args.seed)':             'optuna.samplers.TPESampler(seed=args.seed, n_startup_trials=4)',
}
for a, b in rep.items():
    if a in s:
        s = s.replace(a, b)
    else:
        print("NOT FOUND (already changed?):", a)

open(p, "w").write(s)
import ast; ast.parse(s); print("tune.py updated, syntax OK")
```
